## #03. 파생변수 생성

### 1. 객실 유무


```python
df4 = df3.copy()

# Cabin 컬럼이 결측치가 아닌 경우 객실있음, 결측치인 경우 객실 없음으로 파생변수 생성
df4["HasCabin"] = df4["Cabin"].notnull()

# Cabin 컬럼은 결측치가 많으므로 제거
df4.drop(columns=["Cabin"], inplace=True)

# HasCabin 컬럼을 category 타입으로 변환
df4 = my_qtcheck.set_type(df4, as_category=["HasCabin"])

df4.head()
```

### 2. 동승 가족 수 계산


```python
df5 = df4.copy()

# 새로운 파생 변수 생성 (동승 가족 수 = 동승한 형제·배우자 수 + 동승한 부모·자녀 수 + 본인)
df5["FamilySize"] = df5["SibSp"] + df5["Parch"] + 1

# 결과 확인
df5.head()
```

### 3. 신분 파생변수 추가 (1) - 이름에서 호칭 분리


```python
# 이름에서 호칭 추출 : "성, 호칭. 이름" 형식이므로 쉼표와 마침표 사이를 뽑는다
title_raw = df5["Name"].str.extract(r',\s*([^\.]+)\.', expand=False)
title_raw = title_raw.str.strip()

title_raw.value_counts().rename("빈도").to_frame().T
```

### 4. 신분 파생변수 추가 (2) - 의미별 호칭 정의


```python
title_map = {
    # 성인 남성
    "Mr": "Mr", "Don": "Mr", "Sir": "Mr", "Jonkheer": "Mr",

    # 미혼 여성
    "Miss": "Miss", "Mlle": "Miss", "Ms": "Miss",

    # 기혼 여성
    "Mrs": "Mrs", "Mme": "Mrs", "Dona": "Mrs",
    "Lady": "Mrs", "the Countess": "Mrs",

    # 남자아이
    "Master": "Master",

    # 전문직·군인·성직자
    "Dr": "Officer", "Rev": "Officer", "Col": "Officer",
    "Major": "Officer", "Capt": "Officer",
}

title_map
```

### 5. 신분 파생변수 추가 (3) - 호칭 파생변수 추가 (신분을 의미)


```python
df6 = df5.copy()

# Title 변수 추가
df6["Title"] = title_raw.map(title_map)

# 매핑 누락 점검
print("Title 매핑 누락 :", df6["Title"].isna().sum(), "건\n")

# 문자열이므로 명목형으로 선언
df6 = my_qtcheck.set_type(df6, as_category=["Title"])

# 이름 컬럼 제거
df6 = df6.drop(columns=["Name"])

df6.head()
```

### 6. 생성된 신분 파생 변수의 분포 확인


```python
# 신분에 따른 성별 분포
display(crosstab(df6["Title"], df6["Sex"]))

# 신분에 따른 나이 분포 피벗 테이블
display(
    pivot_table(
        df6,
        index="Title",
        values="Age",
        aggfunc=["count", "median", "min", "max"],
        observed=False,
    )
)
```

## #04. 나이 변수에 대한 결측치 처리

### 1. 신분별로 결측치 대체 기준 세우기


```python
# 신분별 나이 분포 요약
age_stats = df6.groupby("Title", observed=True)["Age"].agg(
    인원="size",
    결측=lambda s: s.isna().sum(),
    mean="mean",
    median="median",
    skew="skew",
)

# 대체값 판단 규칙 : |왜도| <= 0.5 -> 평균, 그 밖에는 중앙값
age_stats["대체기준"] = (
    age_stats["skew"].abs() <= 0.5
).map({True: "mean", False: "median"})

# 대체값
age_stats["대체값"] = np.where(
    age_stats["대체기준"] == "mean",
    age_stats["mean"],
    age_stats["median"]
)

age_stats
```

### 2. 나이 결측치 처리


```python
df7 = df6.copy()

# 신분(Title)별로 결측 Age를 집단별 대표값으로 대체
for title in age_stats.index:
    mask = df7["Age"].isna() & (df7["Title"] == title)
    df7.loc[mask, "Age"] = age_stats.loc[title, "대체값"]

my_qtcheck.check_missing_values(df7)
```
